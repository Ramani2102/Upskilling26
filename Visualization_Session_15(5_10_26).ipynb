```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/placementdata.csv")
```

```python
df.columns = df.columns.str.strip()
```

```python
data = df.head(100).copy()
```

```python
print("First 100 Records:")
print(data)
```

```python
# Step 3: Line Chart - CGPA of First 100 Students
plt.figure(figsize=(10, 5))
plt.plot(data["StudentID"], data["CGPA"], marker="o")
plt.title("CGPA of First 100 Students")
plt.xlabel("Student ID")
plt.ylabel("CGPA")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Step 4: Bar Chart - Placement Status
placement_counts = data["PlacementStatus"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(placement_counts.index.astype(str), placement_counts.values)
plt.title("Placement Status of First 100 Students")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
# Step 5: Pie Chart - Placement Status Percentage
plt.figure(figsize=(7, 7))
plt.pie(
    placement_counts.values,
    labels=placement_counts.index.astype(str),
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Placement Status Percentage")
plt.tight_layout()
plt.show()
```

```python
# Step 6: Scatter Plot - CGPA vs Aptitude Test Score
plt.figure(figsize=(8, 5))
plt.scatter(data["CGPA"], data["AptitudeTestScore"])
plt.title("CGPA vs Aptitude Test Score")
plt.xlabel("CGPA")
plt.ylabel("Aptitude Test Score")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Step 7: Histogram - CGPA Distribution
plt.figure(figsize=(8, 5))
plt.hist(data["CGPA"].dropna(), bins=10, edgecolor="black")
plt.title("Distribution of CGPA")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
