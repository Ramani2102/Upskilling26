```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {"group_a" : [45, 48, 50, 52, 49, 51, 47, 50],
"group_b" : [60, 62, 65, 63, 61, 64, 66, 95]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
print("MIN of group A : ", df['group_a'].min())
print("MAX of group A : ", df['group_a'].max())
print("Mean of group A : ", df['group_a'].mean())
print("Median of group A : ", df['group_a'].median())
print("Std dev of group A : ", df['group_a'].std())
```

```python
print("MIN of group B : ", df['group_b'].min())
print("MAX of group B : ", df['group_b'].max())
print("Mean of group B : ", df['group_b'].mean())
print("Median of group B : ", df['group_b'].median())
print("Std dev of group B : ", df['group_b'].std())
```

```python
# Calculate Q1, Q3, and IQR for group_a
Q1_a = df['group_a'].quantile(0.25)
Q3_a = df['group_a'].quantile(0.75)
IQR_a = Q3_a - Q1_a

print(f"Group A - Q1: {Q1_a}, Q3: {Q3_a}, IQR: {IQR_a}")
```

```python
# Calculate Q1, Q3, and IQR for group_b
Q1_b = df['group_b'].quantile(0.25)
Q3_b = df['group_b'].quantile(0.75)
IQR_b = Q3_b - Q1_b

print(f"Group B - Q1: {Q1_b}, Q3: {Q3_b}, IQR: {IQR_b}")
```

```python
# Calculate boundaries for outliers
lower_bound_a = Q1_a - 1.5 * IQR_a
upper_bound_a = Q3_a + 1.5 * IQR_a
print(f"Lower bound for group A: {lower_bound_a}")
print(f"Upper bound for group A: {upper_bound_a}")
lower_bound_b = Q1_b - 1.5 * IQR_b
upper_bound_b = Q3_b + 1.5 * IQR_b
print(f"Lower bound for group B: {lower_bound_b}")
print(f"Upper bound for group B: {upper_bound_b}")
# Identify outliers
outliers_a = df['group_a'][(df['group_a'] < lower_bound_a) | (df['group_a'] > upper_bound_a)]
outliers_b = df['group_b'][(df['group_b'] < lower_bound_b) | (df['group_b'] > upper_bound_b)]

print("Outliers in Group A:", outliers_a.tolist())
print("Outliers in Group B:", outliers_b.tolist())

# Create a box plot
plt.figure(figsize=(8, 6))
sns.boxplot(data=df[['group_a', 'group_b']])
plt.title('Box Plot of Group A and Group B')
plt.ylabel('Values')
plt.show()
```
